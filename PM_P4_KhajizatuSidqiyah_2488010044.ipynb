```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

```python
# langkah 1
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())

df
```

```python
# langkah 2
X = df.drop(columns=['membeli'])
y = df['membeli']

print(X)
print(y)
```

```python
# langkah 3
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})

print(X)
print (y)
```

```python
# langkah 4
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)

print("Jumlah data X_train:", len(X_train))
print("Jumlah data X_test :", len(X_test))
print("Jumlah data y_train:", len(y_train))
print("Jumlah data y_test :", len(y_test))
```

```python
# langkah 5
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage

print(X_train[num].head())
print(X_test[num].head())
```

```python
# langkah 6

print("Jumlah nilai hilang pada X_train:")
print(X_train.isnull().sum())

print("\nJumlah nilai hilang pada X_test:")
print(X_test.isnull().sum())

print("\nTipe data X_train:")
print(X_train.dtypes)

print("\nStatistik X_train setelah scaling:")
print(X_train.describe())
```

```python
# soal no 1

# Membuat salinan data awal
df_mean = pd.DataFrame(data)

# Menghitung mean dan median usia
mean_usia = df_mean['usia'].mean()
median_usia = df_mean['usia'].median()

print("Mean usia   :", mean_usia)
print("Median usia :", median_usia)

# Imputasi usia menggunakan mean
df_mean['usia'] = df_mean['usia'].fillna(mean_usia)

# Pendapatan tetap menggunakan median
df_mean['pendapatan'] = df_mean['pendapatan'].fillna(
    df_mean['pendapatan'].median()
)

print("\nData setelah imputasi mean:")
print(df_mean)
```

imputasi menggunakan mean mengisi nilai usia yang hilang dengan rata-rata usia, sedangkan pada Bagian C digunakan median. Mean lebih mudah dipengaruhi oleh nilai ekstrem, sedangkan median lebih tahan terhadap nilai ekstrem.

```python
# soal no 2

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

# Membuat X dan y
X_mm = df.drop(columns=['membeli']).copy()
y_mm = df['membeli'].copy()

# Encoding pendidikan
X_mm['pendidikan'] = X_mm['pendidikan'].map({
    'SMA': 0,
    'S1': 1,
    'S2': 2
})

# One-hot encoding kota
X_mm = pd.get_dummies(
    X_mm,
    columns=['kota'],
    dtype=int
)

# Encoding label
y_mm = y_mm.map({
    'Tidak': 0,
    'Ya': 1
})

# Split data
X_train_mm, X_test_mm, y_train_mm, y_test_mm = train_test_split(
    X_mm,
    y_mm,
    test_size=0.3,
    random_state=42
)

# Scaling dengan MinMaxScaler
num = ['usia', 'pendapatan', 'pendidikan']

scaler_mm = MinMaxScaler()

X_train_mm[num] = scaler_mm.fit_transform(X_train_mm[num])
X_test_mm[num] = scaler_mm.transform(X_test_mm[num])

print("Hasil X_train setelah MinMaxScaler:")
print(X_train_mm)

print("\nDescribe X_train:")
print(X_train_mm.describe())
```

```python
# soal no 3

# Membuat salinan data
df_kategori = df.copy()

# Menambahkan kolom pengalaman
df_kategori['pengalaman'] = [
    'Pemula',
    'Menengah',
    'Pemula',
    'Berpengalaman',
    'Menengah',
    'Berpengalaman',
    'Menengah',
    'Pemula',
    'Pemula',
    'Berpengalaman'
]

print("Data sebelum one-hot encoding:")
print(df_kategori)

# One-hot encoding kolom pengalaman
df_kategori = pd.get_dummies(
    df_kategori,
    columns=['pengalaman'],
    dtype=int
)

print("\nData setelah one-hot encoding:")
print(df_kategori)
```

fit_transform() digunakan pada data latih karena scaler perlu mempelajari parameter dari data latih, seperti mean dan standar deviasi pada StandardScaler. Setelah parameter diperoleh, data uji hanya menggunakan transform() dengan parameter yang sama.

1. Mengapa urutan 'split dulu, baru scaling' penting?
2. Kapan label encoding dan kapan one-hot encoding?
3. Apa perbedaan normalisasi dan standardisasi?

jawaban:
1. data harus dibagi menjadi data latih dan data uji terlebih dahulu sebelum melakukan scaling, untuk mencegah kebocoran data. scaler hanya boleh mempelajari di-fit data latih, kemudian digunakan untuk melakukan transform pada data latih dan data uji. jika scaling digunakan sebelum pembagian data, informasi dari data uji dapat ikut masuk ke proses pelatihan.

2. label encoding digunakan untuk data kategorikal yang memiliki urutan atau tingkatan atau bisa disebut ordinal, seperti SMA, S1, dan S2. sedangkan one-hot encoding digunakan untuk data kategorikal yang tidak memiliki urutan, seperti Bandung, Jakarta, dan Surabaya.

3. normalisasi (Min-Max) mengubah nilai data dalam rentang tertentu, biasanya 0-1, sehingga memiliki batas yang jelas tetapi sensitif terhadap outlier. sedangkan standardisasi (z-scrore) mengubah data sehingga memiliki mean mendakati 0 dan standar deviasi 1, secara tidak memiliki batas rentang tertentu.
